```python
import collections
import pandas as pd
import xarray as xr
import numpy as np
from pathlib import Path
```

```python
src_dir = Path('/gpfs/work2/0/prjs1027/tfrecords/run4')
```

```python
nc_files = list((src_dir / 'min_max_files').glob('*H*.nc'))
```

```python
min_max_values = {}
for var in xr.open_dataset(nc_files[0]).data_vars:
    min_max_values[var] = np.array([np.inf, -np.inf])
```

```python
max_hs = []
for ds_path in nc_files:
    ds = xr.open_dataset(ds_path)
    max_hs.append(ds['hs'][1].values)
```

```python
max_hs
```

```python
for ds_path in nc_files:
    ds = xr.open_dataset(ds_path)
    for var in ds.data_vars:
        min_max_values[var][0] = np.min([min_max_values[var][0], ds[var][0]])
        min_max_values[var][1] = np.max([min_max_values[var][1], ds[var][1]])
```

```python
min_max_values_df = pd.DataFrame(min_max_values)
min_max_values_df.index = ['minimum', 'maximum']
min_max_values_ds = xr.Dataset.from_dataframe(min_max_values_df)
```

```python
min_max_values_ds.to_netcdf(src_dir / 'min_max_files/min-max-values-all.nc')
```
