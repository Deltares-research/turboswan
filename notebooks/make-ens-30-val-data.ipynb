```python
for ens in range(26, 31):
    ds_path = src_path / f'ens{ens}'
    datasets = list(ds_path.glob('*/output/*.nc'))

    datasets.sort()

    ds = xr.concat([xr.open_dataset(data) for data in datasets], dim='time')
    ds = ds.drop_duplicates(dim="time", keep="last")
    ds.to_netcdf(write_dir / f'validation_ens{ens}.nc')
```
